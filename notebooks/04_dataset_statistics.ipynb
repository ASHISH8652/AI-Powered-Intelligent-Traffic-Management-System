# Module 4 – Dataset Preprocessing & Preparation

```python
from pathlib import Path
import cv2
import pandas as pd
```

```python
DATASET_PATH = Path(r"C:\AI-Powered-Intelligent-Traffic-Management-System\datasets\DETRAC_Upload")

TRAIN_IMAGES = DATASET_PATH / "images/train"
TRAIN_LABELS = DATASET_PATH / "labels/train"
```

```python
image_files = sorted(TRAIN_IMAGES.glob("*.jpg"))
label_files = sorted(TRAIN_LABELS.glob("*.txt"))

matched = 0
missing = 0

for img in image_files:

    label = TRAIN_LABELS / (img.stem + ".txt")

    if label.exists():
        matched += 1
    else:
        missing += 1

print("="*60)
print("IMAGE-LABEL VERIFICATION")
print("="*60)

print("Matched :", matched)
print("Missing :", missing)
```

```python
corrupted = []

for img in image_files:

    image = cv2.imread(str(img))

    if image is None:
        corrupted.append(img.name)

print("="*60)
print("CORRUPTED IMAGES")
print("="*60)

print("Total :", len(corrupted))
```

```python
invalid_labels = []

for label in label_files:

    with open(label) as f:

        for line in f:

            values = line.split()

            if len(values) != 5:

                invalid_labels.append(label.name)
print("="*60)
print("INVALID LABEL FILES")
print("="*60)

print(len(invalid_labels))
```

```python
errors = []

for label in label_files:

    with open(label) as f:

        for line in f:

            cls, x, y, w, h = map(float, line.split())

            if not (
                0 <= x <= 1 and
                0 <= y <= 1 and
                0 <= w <= 1 and
                0 <= h <= 1
            ):

                errors.append(label.name)

print("="*60)
print("COORDINATE ERRORS")
print("="*60)

print(len(errors))
```

```python
classes = set()

for label in label_files:

    with open(label) as f:

        for line in f:

            cls = int(line.split()[0])

            classes.add(cls)

print("="*60)
print("CLASSES FOUND")
print("="*60)

print(sorted(classes))
```

```python
filenames = [img.name for img in image_files]

duplicates = len(filenames) - len(set(filenames))

print("="*60)
print("DUPLICATE FILES")
print("="*60)

print("Duplicates :", duplicates)
```

```python
report = pd.DataFrame({

    "Check":[
        "Image-Label Match",
        "Corrupted Images",
        "Invalid Labels",
        "Coordinate Errors",
        "Duplicate Images"
    ],

    "Result":[
        matched,
        len(corrupted),
        len(invalid_labels),
        len(errors),
        duplicates
    ]

})

report
```
