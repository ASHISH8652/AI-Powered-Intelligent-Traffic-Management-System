# Dataset Exploration

This notebook explores and validates the UA-DETRAC dataset before model training.

```python
import os
import yaml
from pathlib import Path
import pandas as pd
```

```python
DATASET_PATH = Path(r"C:\AI-Powered-Intelligent-Traffic-Management-System\datasets\DETRAC_Upload")

TRAIN_IMAGES = DATASET_PATH / "images" / "train"
VAL_IMAGES = DATASET_PATH / "images" / "val"

TRAIN_LABELS = DATASET_PATH / "labels" / "train"
VAL_LABELS = DATASET_PATH / "labels" / "val"

DATA_YAML = DATASET_PATH / "data.yaml"
```

```python
print("=" * 60)
print("DATASET VALIDATION")
print("=" * 60)

folders = [
    TRAIN_IMAGES,
    VAL_IMAGES,
    TRAIN_LABELS,
    VAL_LABELS
]

for folder in folders:
    if folder.exists():
        print(f"✓ {folder}")
    else:
        print(f"✗ Missing: {folder}")
```

```python
train_images = list(TRAIN_IMAGES.glob("*.jpg"))
val_images = list(VAL_IMAGES.glob("*.jpg"))

train_labels = list(TRAIN_LABELS.glob("*.txt"))
val_labels = list(VAL_LABELS.glob("*.txt"))

print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print(f"Training Images    : {len(train_images):,}")
print(f"Training Labels    : {len(train_labels):,}")
print()

print(f"Validation Images  : {len(val_images):,}")
print(f"Validation Labels  : {len(val_labels):,}")
```

```python
total_images = len(train_images) + len(val_images)

train_percent = (len(train_images) / total_images) * 100
val_percent = (len(val_images) / total_images) * 100

print("=" * 60)
print("DATASET SPLIT")
print("=" * 60)

print(f"Total Images : {total_images:,}")
print(f"Training     : {train_percent:.2f}%")
print(f"Validation   : {val_percent:.2f}%")
```

```python
print("=" * 60)
print("CHECKING MISSING LABELS")
print("=" * 60)

missing_labels = []

for img in train_images:
    label = TRAIN_LABELS / (img.stem + ".txt")

    if not label.exists():
        missing_labels.append(img.name)

print(f"Missing Labels : {len(missing_labels)}")

if len(missing_labels) > 0:
    print(missing_labels[:10])
else:
    print("✓ All training images have labels.")
```

```python
print("=" * 60)
print("CHECKING MISSING IMAGES")
print("=" * 60)

missing_images = []

for label in train_labels:
    image = TRAIN_IMAGES / (label.stem + ".jpg")

    if not image.exists():
        missing_images.append(label.name)

print(f"Missing Images : {len(missing_images)}")

if len(missing_images) > 0:
    print(missing_images[:10])
else:
    print("✓ All labels have corresponding images.")
```

```python
print("=" * 60)
print("SAMPLE TRAIN IMAGES")
print("=" * 60)

for img in train_images[:10]:
    print(img.name)
```

```python
print("=" * 60)
print("DATA.YAML CONTENTS")
print("=" * 60)

if DATA_YAML.exists():
    with open(DATA_YAML, "r") as file:
        config = yaml.safe_load(file)

    print(config)
else:
    print("data.yaml not found.")
```

```python
summary = pd.DataFrame({
    "Category": [
        "Training Images",
        "Validation Images",
        "Training Labels",
        "Validation Labels",
        "Total Images"
    ],
    "Count": [
        len(train_images),
        len(val_images),
        len(train_labels),
        len(val_labels),
        total_images
    ]
})

summary
```

```python
print("=" * 60)
print("FINAL DATASET HEALTH CHECK")
print("=" * 60)

if (
    len(train_images) == len(train_labels)
    and len(val_images) == len(val_labels)
):
    print("✓ Dataset is VALID")
    print("✓ Images and labels are matched")
    print("✓ Dataset is ready for visualization")
    print("✓ Dataset is ready for YOLO training")
else:
    print("✗ Dataset has inconsistencies")
```
